```python
def pivot(df, index, id = "id"):
    df = pd.read_csv(f"./data/extracted/csv/{df}.csv", encoding = "utf-8").fillna("")
    df["id"] = df[f"{id}"] 
    columns = df.drop(columns = [index, "id"])
    values = df.columns
    
    df["group"] = df.groupby([f"{id}"]).cumcount() + 1
        
    wide = (
        df.pivot(
            index= "id",
            columns="group",
            values= values
        )
    ).reset_index()
    
    
    wide.columns = [
        f"{name}{group}" for name, group in wide.columns
    ]
        
    
    #    df = df.drop(columns = f"{index}2").fillna("")
    df = df.reset_index()
    return df
```
